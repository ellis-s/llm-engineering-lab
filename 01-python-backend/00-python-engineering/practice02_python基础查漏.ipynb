## 集合

> 它是“只有键、没有值的字典"，对标 JS 的 Set，是去重和关系运算的利器。

**核心特性**：无序 + 元素必须可哈希



````python
s1 = {1, 2, 3}  # ① 字面量
s2 = set([1, 2, 2, 3, 3])  # ② 从列表转 → {1, 2, 3}，重复自动消失
s3 = set("hello")  # ③ 从字符串 → {'h', 'e', 'l', 'o'}
s4 = set()  # ✅ 空集合只能这么写！

s5 = {}  # ❌ 这是空字典！！不是集合！
type(s5)  # dict —— JS 转过来的头号坑
````


### 常用方法
```python
s = {1, 2, 3}

s.add(4)              # 增（单个）
s.update([5, 6])      # 批量增 —— 又见 update！和 dict.update 同款逻辑
s.discard(99)         # 删：不存在也不报错
s.remove(99)          # 删：不存在就 KeyError —— 类似 dict 的 del
s.pop()               # 随机弹出一个（无序，所以没有"末尾"概念）
s.clear()             # 清空

2 in s                # True —— 成员判断是集合的招牌优势！
```

#### 3.2【查找（读取）】变量名称的顺序

遵循 `LEGB` 规则。 查找顺序是： L -> E -> G -> B

1. L: Local       当前函数
2. E: Enclosing   外层函数（如果存在）
3. G: Global      当前模块
4. B: Built-in    内置名称


另外，需要注意一点，由于`LEGB`规则，在类的函数体中，不加 `self.` 是无法直接访问到类中的属性。比如：
```python
x = 1

class Student:
  x = 2

  def get_x():
    print(x)

student = Student()
student.get_x()  # 1

```


#### 3.3【赋值】变量

1. 默认情况，创建局部变量。
2. 修改全局变量，需要先声明 global
3. 修改外层函数变量，需要先声明 nonlocal


**关于 nonlocal**

nonlocal 用于嵌套函数，找到最近的、已经存在的外层函数作用域中的绑定，并修改它。

如果没有合适的外层函数绑定，使用 nonlocal 会报 SyntaxError。

是实现`函数闭包`的重要支持。


````python
a = 1
num=[]


def default_set_variable():
  a = 2
  return a

default_set_variable()
print(f'默认情况，外部 a = {a}')


def change_global_variable():
  '''
  修改函数外变量，需要声明 global。 否则会报错
  '''
  global a
  a  = 2

change_global_variable()
print(f'显示声明修改外部变量后, 外部 a = {a}')



def outer():
  a = 11
  
  def inner():
    nonlocal a
    a = 22

  inner()
  print(f'修改外层函数的变量后, a = {a}')

outer()




x = 1
class Student:
    x = 2

    def get_x(self):
      return x

    def get_self_x(self):
      return self.x
    
    def get_student_x(self):
      return Student.x

student = Student()
print(student.get_x())  # 1
print(student.get_self_x())
print(student.get_student_x())
````

### 4. 函数传参顺序

1. * 和 *args 不能同时存在。 * 相当于分割线，规定后面的参数必须使用关键字传参。 *args 既可以有分割线的作用，又可以收集多余的位置参数。
2. /: 相当于分割线，规定左侧参数必须使用位置参数传参。 
3. 在 / 和 * 之间的参数，既可以位置传参，又可以使用关键字传参。

````python

def fn(name, /, age=18, *, gendar, **kwargs ):
    print(name)
    print(age)
    # print(args)
    print(gendar)
    print(kwargs)


fn('xiaoming', 20, gendar='woman', hobby='打篮球')
````

## 类

### 私有属性 & 私有方法

只能在类内部使用。

以 `__`（两个下划线）开头的变量，被视为是私有属性。

````python
class Person:
  __money = 100

  def __show_money(self):
    print(self.__money)
    return self.__money

p1 = Person()

p1.__money
p1.__show_money()
````
