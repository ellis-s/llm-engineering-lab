````python
def get_most_benifit(prices: list[int]):

    l = len(prices)

    if l < 2:
        return None 

    """
      1. 以列表首元素为当前买入价，以第二个元素为全局卖出价，二者差为全局利润。
      2. 循环遍历索引 1 至末尾。
      3. 在循环中，用当前元素与当前买入价计算当前利润。
      4. 若当前利润大于全局利润，更新全局利润与全局卖出价。
      5. 若当前买入价大于当前元素，更新当前买入价。
      6. 最终返回全局买入与全局卖出：全局买入 = 全局卖出 - 全局利润。
    """

    cur_buy_i = 0
    cur_sell_i = 1

    res = prices[1] - prices[0]

    for i in range(2, len(prices)):
        new_res = prices[i] - prices[cur_buy_i]

        old_sell_i = cur_sell_i

        #  新的值更大
        if new_res > res:
          res = new_res
          cur_sell_i = i
          
          # 获取最低的买入价
          if prices[cur_buy_i] > prices[old_sell_i]:
            cur_buy_i = old_sell_i

        else:
            # 新的值更小
            second_res = prices[i] - prices[cur_sell_i]
            if second_res > res:
                res = second_res
                cur_buy_i = cur_sell_i
                cur_sell_i = i

    return prices[cur_buy_i], prices[cur_sell_i], res


# get_most_benifit([8, 4, 12, 9, 20, 1])
# get_most_benifit([8, 6, 5, 4, 3, 2, 1])
get_most_benifit([10, 9, 16, 17, 19, 23])
````

````python
def backpack01(max_weight:int, item: dict[int, int]):
    dp = [0] * (max_weight + 1)

    for weight, value in item.items():
        for i in range(max_weight, weight - 1, -1):
            dp[i] = max(dp[i - weight] + value, dp[i])
        print(dp)


    return dp[max_weight]

d = {
  2: 6,
  3: 10,
  4: 12,
  5: 15
}


backpack01(10, d)
````

````python
def backpack_unlimited(max_weight: int, item: dict[int, int]):
    dp = [0] * (max_weight + 1)

    for weight, value in item.items():
        for i in range(weight, max_weight + 1):
            dp[i] = max(dp[i - weight] + value, dp[i])
        print(dp)

    return dp[max_weight]


d = {2: 6, 3: 10, 4: 12, 5: 15}

backpack_unlimited(10, d)
````

````python
def coin_change(coins: list[int], amount: int) -> int:
    dp = [0] * (amount + 1)
    dp[0] = 1
    print(dp)

    for coin in coins:
        for i in range(coin, amount + 1):
            dp[i] += dp[i - coin]

        print(dp)

    return dp[amount]


print(coin_change([1, 2, 5], 5))  # 4
````

````python
def climb_stairs(n: int) -> int:
    dp = [0] * (n + 1)

    dp[0] = 1
    dp[1] = 1

    for i in range(2, n + 1):
        # 在这里补充状态转移
        dp[i] = dp[i - 1] + dp[i - 2]

    print(dp)
    
    return dp[n]

climb_stairs(5)
````

````python
import math


def maxSubArray(nums: list[int]) -> int:
    # 错误思想 1
    # 1. dp[i][j] 表示连续从i～j数字的最大和。
    # 2. dp[0] 表示负无穷大
    # 3. dp[i][j] = dp[i][j-1] + value[j]

    # 错误思想 2
    # dp[i] 表示连续 i 个元素中的最大和
    # dp[0] 表示负无穷大
    # 最外层 for 循环 nums, 表示行
    # 最内层 for 循环 nums.length , 表示列
    # dp[i] = dp[i]

    """
    dp[i] 表示以 i 结尾时，前面子数组的最大和。
    dp[i] 时，有两种方案
        1. 前面所有子数组 + 当前第 i 个元素
        2. 从第 i 个元素开始重新计算
    所以： dp[i] = max(dp[i - 1] + value[i], value[i])

    最终取 max(dp) 即可。
    """

    # dp = list(nums)

    # for i in range(1, len(nums)):
    #     dp[i] = max(dp[i - 1] + dp[i], dp[i])
    #     print(dp)

    # return max(dp)

    # 以当前元素结尾的最大子数组和
    current_i_max_sum = -math.inf
    # 遍历到目前为止，所有位置的最大子数组和
    final_max_sum = current_i_max_sum
    

    for _, value in enumerate(nums):
        current_i_max_sum = max(current_i_max_sum + value, value)

        final_max_sum = max(final_max_sum, current_i_max_sum)
        print(current_i_max_sum)

    return final_max_sum


maxSubArray([-2, 1, -3, 4, -1, 2, 1, -5, 4])
````
