
# NumPy

NumPy 提供了数组对象。


针对于接下来要学习的 AI 大模型应用开发和 Agent 开发领域，我只需要掌握以下内容。

**NumPy 不需要学到科学计算专家的程度，但必须掌握数组、维度、向量运算、矩阵运算和数据处理。**

1. LLM 应用开发：NumPy 常用于 Embedding 向量处理、相似度计算、数据预处理、评估指标和实验分析。
2. Agent 开发：NumPy 主要用于工具返回的数据处理、结构化数值计算、排序、评分及优化算法。Agent 的工具调用、状态管理、规划与记忆本身并不依赖 NumPy。
3. 模型训练与推理：通常以 PyTorch 张量为主，但 NumPy 与 Tensor 之间的数据转换非常常见。


```python
import numpy as np

np.__version__
```

```python
a1 = np.array([[1, 2, 3], [4, 5, 6]])

print(a1.ndim) # 几个轴
print(a1.shape) # 每个轴分别有多长
print(a1.size) # 总共有多少元素
print(a1.dtype)



print(a1.reshape(1,6)) # 在不改变 size 的情况下，修改形状。

print(a1.T) # 矩阵的转置


```

```python
a = np.array([[1, 2], [3, 4]])
b = np.array([[5, 6], [7, 8]])

print(a * b)  # 对应位置相乘
print(a @ b)  # 矩阵乘法
```
