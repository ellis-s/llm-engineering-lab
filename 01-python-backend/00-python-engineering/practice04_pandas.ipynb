## 接下来，跟着 GPT 给的内容进行练习

````python
import pandas as pd

df = pd.DataFrame(
    {
        "name": ["Alice", "Bob", "Charlie", "David", "Alice"],
        "age": [25, 30, 28, None, 25],
        "score": [90, 85, 95, 70, 90],
        "city": ["Shanghai", "Beijing", "Shanghai", "Shenzhen", "Shanghai"],
        "category": ["A", "B", "A", "B", "A"],
    }
)

print(df)
````

### Series 练习

练习目标：
- 分别打印行数、列数、列名和数据类型。
- 观察 df["score"] 与 df[["score"]] 的区别。
- 使用 describe() 查看平均值、标准差、最小值和最大值。

````python
row, col = df.shape


print(f'行数 row = {row}')
print(f'列数 col = {col}')
print(df.columns)
print(df.dtypes)

print('=' * 20)
print(df["score"], type(df['score'])) # 返回的是一个 Series
print(df[['score']], type(df[['score']])) # 返回一个 DataFrame
````

````python
des = df.describe()
print(des, type(des))


des['age']['mean']
````

````python
print(df)
# 1. 获取 Alice 的分数
print(df[df["name"] == "Alice"].loc[0]["score"])
print(df[df["name"] == "Alice"]['score'])
print(df.loc[df['name'] == 'Alice', 'score'])

````

````python
# 2. 获取第 2 个元素
df.iloc[1]
````

````python
# 3. 找出分数大于等于 90 的元素
df[df["score"] >= 90]

````

````python
# 4. 找出分数在 80 到 95 之间的元素
df[df["score"].between(80, 95)]

df[(df['score'] > 80) & (df['score']<95)]
````

````python
# 5. 获取分数最高的 3 个元素
df.nlargest(3, columns="score")
````

````python
# 6. 统计每种分数出现的次数
df["score"].value_counts()
````

### dataframe 条件筛选和排序

````python
# 1. 筛选年龄大于 25 岁的人
df[df['age'] > 25]
````

````python
# 2. 筛选城市为 Shanghai 且分数 >= 90 的人
df[(df['city'] == 'Shanghai') & (df['score'] >= 90)]
````

````python
# 3. 筛选分数属于 85、90、95 的记录
df[(df["score"] == 85) | (df["score"] == 90) | (df["score"] == 95)]

# 写法 2

df[df['score'].isin([85,90,95])]
````

````python
# 4. 按分数从高到低排序
df.sort_values('score', ascending=False)
````

````python
# 5. 找出分数最高的 3 条记录
df.nlargest(3, 'score')
````

````python
# 6. 按年龄升序、分数降序进行多列排序
df_1 = pd.DataFrame(
    {
        "name": ["Alice", "Bob", "Charlie", "David", "Ellis"],
        "age": [25, 30, 28, None, 25],
        "score": [90, 85, 95, 70, 92],
        "city": ["Shanghai", "Beijing", "Shanghai", "Shenzhen", "Shanghai"],
        "category": ["A", "B", "A", "B", "A"],
    }
)
df_1.sort_values(["age", "score"], ascending=[True, False])
````

### 修改 dataframe 
```python
# 1. 新增一列 passed，分数 >= 60 为 True

# 2. 新增一列 score_plus_5，分数增加 5 分

# 3. 将 score 列重命名为 points

# 4. 删除 category 列

# 5. 将 age 转换为合适的数据类型

# 6. 修改某一行某个单元格的值
```

````python
# 1. 新增一列 passed，分数 >= 80 为 True
df['passed'] = df['score'] >= 80
df
````

````python
# 2. 新增一列 score_plus_5，分数增加 5 分
df['score_plus_5'] = df['score'] + 5
df
````

````python
# 3. 将 score 列重命名为 points
df.rename(columns={"score": "points"}, inplace=[True])
df

````

````python
# 4. 删除 category 列
df.drop(columns=['category'])


````

````python
# 5. 将 age 转换为合适的数据类型
df['age'] = df["age"].astype("Int64")
df

````

````python
# 6. 修改某一行某个单元格的值
df1 = df.copy()
df1.loc[3, 'age'] = 24
df1
# df
````
